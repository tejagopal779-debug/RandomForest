```python
import pandas as pd
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.model_selection import train_test_split
import numpy as np
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
df=pd.read_csv(r"C:\Users\DELL\Downloads\loan\customer_churn_data.csv")
```

```python
df
```

```python
x=df.drop("Churn",axis=1)
y=df["Churn"]
```

```python
x_num=x.select_dtypes(include="number")
x_cat=x.select_dtypes(exclude="number")
```

```python
d=StandardScaler()
x_num=d.fit_transform(x_num)
```

```python
p=OneHotEncoder()
x_cat=p.fit_transform(x_cat).toarray()
```

```python

x_final=np.hstack([x_num,x_cat])
```

```python
x_final
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
x_tri, x_test, y_tra, y_tse=train_test_split(x_final,y,test_size=0.15) 
```

```python
k=RandomForestClassifier()
w=k.fit(x_tri,y_tra)
```

```python
per=k.predict(x_test)
```

```python
acc=accuracy_score(y_tse,per)
acc
```
